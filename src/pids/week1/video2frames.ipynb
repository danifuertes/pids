## Week 1 · From videos to frames  🎥 → 🖼️

You have recorded **videos** of the robots. Before they can be labelled, each video has to be split into individual **frames** (still images). That is the goal of this notebook.

**Pipeline**
1. Find every video inside the dataset folder.
2. Read each video and keep one frame every `interval` frames.
3. Save the frames as `.jpg` images, ready for the labelling step.

> 🔑 **The `interval` parameter.** A video has many frames per second, and consecutive frames look almost identical. `interval` sets **how often a frame is kept**: with `interval = 15`, one frame out of every 15 is saved. Use a larger value for fewer, more varied frames; a smaller value for more frames.

Each video gets its own folder: `images/` holds the extracted frames and `labels/` is left empty for the bounding boxes you will add next.

```
dataset/
└── Video_1/
    ├── images/   # extracted frames (.jpg)
    └── labels/   # bounding boxes (added when labelling)
```

### Imports

````python
import os
import cv2
````

### Extract frames

`extract_frames` opens a single video and saves one frame every `interval` frames into an `images/` subfolder.

````python
def extract_frames(video_path, output_dir, interval=15):
    """Split a video into frames, saving one every `interval` frames.

    Args:
        video_path: Path to the input video file.
        output_dir: Folder for this video's output (holds `images/` and `labels/`).
        interval: Save one frame every `interval` frames (higher means fewer frames).
    """
    # Open the video and stop early if it cannot be read
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f'Error: could not open {video_path}')
        return

    # Create the images/ (frames) and labels/ (future annotations) subfolders
    images_dir = os.path.join(output_dir, 'images')
    labels_dir = os.path.join(output_dir, 'labels')
    os.makedirs(images_dir, exist_ok=True)
    os.makedirs(labels_dir, exist_ok=True)

    frame_count = 0
    saved = 0
    while True:
        # Read the next frame; stop at the end of the video
        ret, frame = cap.read()
        if not ret:
            break
        frame_count += 1

        # Save one frame every `interval` frames
        if frame_count % interval == 0:
            image_path = os.path.join(images_dir, f'frame_{frame_count:04d}.jpg')
            cv2.imwrite(image_path, frame)
            saved += 1

    cap.release()
    print(f' -> Frames extracted: {saved}')
    print(f' -> Frames saved in: {images_dir}\n')
````

### Run the extraction

Point `SOURCE_DIR` to the folder with your videos and choose the `INTERVAL`, then run the loop over every video.

````python
path_dataset = '../dataset' # Folder with the recorded videos (dataset at the project root)
interval = 15               # Save one frame every INTERVAL frames
````

````python
# Only process video files, which skips hidden folders such as .ipynb_checkpoints
VIDEO_EXTENSIONS = ('.mp4', '.mov', '.avi', '.mkv')

for video_file in sorted(os.listdir(path_dataset)):
    
    # Skip anything that is not a video file
    if not video_file.lower().endswith(VIDEO_EXTENSIONS):
        continue
    print(f'Processing file: {video_file}')

    # Give each video its own output folder, named after the video
    video_path = os.path.join(path_dataset, video_file)
    sequence_name = os.path.splitext(video_file)[0]
    output_dir = os.path.join(path_dataset, sequence_name)

    # Extract frames
    extract_frames(video_path, output_dir, interval=interval)
````
